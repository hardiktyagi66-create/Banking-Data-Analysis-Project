## 3) Loan_data table

```python
df3 = pd.read_csv('loan_data.csv')
df3.head()
```

### Check null values  and duplicated values

```python
n3 = df3.isnull().sum()
n3
```

#### Handling null values

```python
df3["loan_type"].fillna(df3["loan_type"].mode()[0], inplace=True)
df3["loan_amount"].fillna(df3["loan_amount"].mean(), inplace=True)
df3["interest_rate"].fillna(df3["interest_rate"].mean(), inplace=True)
df3["loan_status"].fillna(df3["loan_status"].mode()[0], inplace=True)
df3["loan_start_date"].fillna(method="ffill", inplace=True)
```

```python
df3.isnull().sum()
```

```python
d3 = df3.duplicated().sum()
d3
```

### Check Data Type

```python
df3.dtypes
```

#### Change data type

```python
df3["loan_start_date"] = pd.to_datetime(df3["loan_start_date"])
```

```python
df3.dtypes
```

```python
df3.to_csv("loan.csv", index=False)
```
